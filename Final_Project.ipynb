```python
import tkinter as tk
from tkinter import ttk
import sqlite3
from tkinter import messagebox
# # Creates the main window of your application.
main = tk.Tk()
main.title("Information App")
main.geometry("400x400")  # window size

popups_closed = 0
# Function for submitting the form
def submit_form():
    name = name_var.get()
    favorite_color = color_var.get()
    favorite_animal = animal_var.get()

    # Displays  greeting here
    if name:
        greeting_label.config(text=f"Hello, {name}!", font=("Arial", 14, "bold"), fg="green")
    else:
        greeting_label.config(text="Please enter your name", font=("Arial", 12), fg="red")
    # Save to database heree  
    save_to_database(name, favorite_color, favorite_animal)
    # Display your favorite color in a popup
    if favorite_color:
        show_color_popup(favorite_color)
    # Displays your favorite animal in popup
    if favorite_animal:
        show_animal_popup(favorite_animal)
        
# Function to show favorite color popup
def show_color_popup(favorite_color):
    popup = tk.Toplevel(main)
    popup.title("Favorite Color")
    popup.geometry("200x100")
    popup_label = tk.Label(popup, text=f"Your favorite color is {favorite_color}!", font=("Arial", 12), fg=favorite_color)
    popup_label.pack(pady=20)
    #popup is closed 
    popup.protocol("WM_DELETE_WINDOW", lambda: on_popup_close(popup))
    
# Function to show favorite animal popup
def show_animal_popup(favorite_animal):
    popup = tk.Toplevel(main)
    popup.title("Favorite Animal")
    popup.geometry("200x100")
    popup_label = tk.Label(popup, text=f"Your favorite animal is a {favorite_animal}!", font=("Arial", 12), fg="black")
    popup_label.pack(pady=20)
    popup.protocol("WM_DELETE_WINDOW", lambda: on_popup_close(popup))


#function to save database 
def save_to_database(name, favorite_color, favorite_animal):
    conn = sqlite3.connect('user_data.db')
    cursor = conn.cursor()
    
    cursor.execute('''CREATE TABLE IF NOT EXISTS user_info (name TEXT, color TEXT, animal TEXT)''')
    cursor.execute('''INSERT INTO user_info (name, color, animal) VALUES (?, ?, ?)''', (name, favorite_color, favorite_animal))
    conn.commit()
    conn.close()
def on_popup_close(popup):
    global popups_closed
    popups_closed += 1
    popup.destroy() #close popup\
    if popups_closed == 2:
        show_success_message()
        
def show_success_message():
    messagebox.showinfo("Success", "Your data has been saved successfully!")
# Function to clear the form inputs 
def clear_form():
    name_var.set("")  
    color_var.set("")  
    animal_var.set("") 
    greeting_label.config(text="") 
    
# Labels and Inputs the User Form here
label_name = tk.Label(main, text="Enter your name:", font=("Arial", 12), fg="blue")
label_name.pack(pady=10)

name_var = tk.StringVar()  
name_entry = tk.Entry(main, textvariable=name_var, font=("Arial", 12))
name_entry.pack(pady=10)
label_color = tk.Label(main, text="Select your favorite color:", font=("Arial", 12), fg="blue")
label_color.pack(pady=10)

color_var = tk.StringVar()  # String variable for color
color_combobox = ttk.Combobox(main, textvariable=color_var, values=["Red", "Green", "Blue", "Purple", "Yellow", "Pink"], font=("Arial", 12))
color_combobox.set("")  # Brings this back to blank
color_combobox.pack(pady=10)
label_animal = tk.Label(main, text="Select your favorite animal:", font=("Arial", 12), fg="blue")
label_animal.pack(pady=10)

# Add the ComboBox for animal selection
animal_var = tk.StringVar()  # String variable for animal
animal_combobox = ttk.Combobox(main, textvariable=animal_var, values=["Dog", "Cat", "Bunny", "Bird", "Capybara", "Chinchilla", "Wolf", "Red panda"], font=("Arial", 12))
animal_combobox.set("")  # Default value to blank
animal_combobox.pack(pady=10)
# Label for greeting here
greeting_label = tk.Label(main, text="", font=("Arial", 14), fg="black")
greeting_label.pack(pady=15)
# Submit and Clear buttons here
submit_button = tk.Button(main, text="Submit", font=("Arial", 12), bg="lightblue", fg="black", command=submit_form)
submit_button.pack(pady=10)

clear_button = tk.Button(main, text="Clear", font=("Arial", 12), bg="lightcoral", fg="black", command=clear_form)
clear_button.pack(pady=10)
# Run the main loop
main.mainloop()
```
